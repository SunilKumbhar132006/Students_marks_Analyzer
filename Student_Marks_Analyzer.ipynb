```python
import numpy as np;
```

 Create Subject Names and Student Marks

```python

subjects = np.array([
"maths","science","English","Computer","physics"
])
marks= np.random.randint(35, 100, (50, 5))
print("student Marks:")
print(marks)
```

inspecting shape,ndim, size,dtype

```python

print("Shape:", marks.shape)
print("Dim:", marks.ndim)
print("total Marks:", marks.size)
print("data_type:", marks.dtype)
print("\nfirst 10 students:")
#slicing
print(marks[:10])
```

```python

print("first 10 Students:")
print(marks[:10])
print("\nLast Subject:")
print(marks[:,-1])
print("\ncolumn ranges:")
print(marks[:5,:3])
```

boolean masks

```python

below_40 = marks< 40
above_90 = marks> 90
print("Marks Below 40:")
print(marks[below_40])
print("Marks Above 90:")
print(marks[above_90])
print("\ncount Below 40:",np.sum(below_40))
print("Count Above 90:",np.sum(above_90))
```

subject wise toppers

```python
print("Subject-Wise highest Marks:")
highest = np.max(marks, axis=0)
print(subjects)
print(highest)
```

pass percentage

```python
print("Pass Percentage:")
passed=marks>=40
pass_percentage= np.mean(passed,axis=0) *100
print(subjects)
print(np.round(pass_percentage,2),"%")
```
